```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import random

# 1. Setup Products Data
products = {
    "Food": {"Adasea": 15, "Wika": 10, "Kawal": 12, "Garad": 10, "Seseme oil small": 11, "Seseme oil large": 21},
    "Children Clothing": {"Kids dress": 45, "Kids track suit": 60},
    "Women Clothing": {"Summer dress": 35, "Fine dress": 120, "Abaya": 70, "Hijab": 10, "Toub": 150},
    "Men Clothing": {"Regular jalabia": 70, "Jeans": 50, "T-shirt": 30, "Shirt": 30},
    "Accessories": {"Imma for men": 25, "Curtains": 150, "African fabric": 50, "Jwellery": 3, "Tagia for men": 10}
}

payments = ["Cash", "Card", "E-Transfer"]
start_date = datetime(2024, 8, 24)
end_date = datetime(2025, 7, 23)
date_range = [start_date + timedelta(days=x) for x in range((end_date - start_date).days + 1)]

data = []

# 2. Generate Daily Transactions
for date in date_range:
    daily_revenue = 0
    # Target 200-350 per day
    target_daily = random.uniform(200, 350)

    while daily_revenue < target_daily:
        cat = random.choice(list(products.keys()))
        prod = random.choice(list(products[cat].keys()))
        price = products[cat][prod]
        qty = random.randint(1, 3) if price < 50 else 1
        total = qty * price

        # Prevent huge overshooting of daily target
        if daily_revenue + total > 400:
            break

        data.append([
            date.strftime("%Y-%m-%d"),
            cat,
            prod,
            qty,
            price,
            total,
            random.choice(payments)
        ])
        daily_revenue += total

# 3. Create DataFrames
df = pd.DataFrame(data, columns=["Date", "Main Category", "Product Name", "Qty", "Unit Price", "Total Sales", "Payment Method"])

# Create Summary
summary_data = {
    "Metric": ["Total Annual Revenue", "Total Transactions", "Average Daily Sales", "Date Range Start", "Date Range End"],
    "Value": [
        f"${df['Total Sales'].sum():,.2f}",
        len(df),
        f"${df['Total Sales'].sum()/len(date_range):,.2f}",
        "2024-08-24",
        "2025-07-23"
    ]
}
summary_df = pd.DataFrame(summary_data)

# 4. Save to Excel
with pd.ExcelWriter("Retail_Sales_Dataset.xlsx") as writer:
    df.to_excel(writer, sheet_name="Daily Sales", index=False)
    summary_df.to_excel(writer, sheet_name="Summary", index=False)

print("Excel file 'Retail_Sales_Dataset.xlsx' has been generated.")
```
